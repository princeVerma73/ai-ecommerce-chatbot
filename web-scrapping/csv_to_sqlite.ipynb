```python
import pandas as pd
import sqlite3

# -------------------------------
# 1. Read CSV using Pandas
# -------------------------------
csv_file = "flipkart_product_data.csv"   # your CSV file
df = pd.read_csv(csv_file)

# -------------------------------
# 2. Connect to SQLite Database
# -------------------------------
db_file = "db.sqlite"   # ✅ exact name as required
conn = sqlite3.connect(db_file)
cursor = conn.cursor()

# -------------------------------
# 3. Create Table (if not exists)
# -------------------------------
create_table_query = """
CREATE TABLE IF NOT EXISTS product (
    product_link TEXT,
    title TEXT,
    brand TEXT,
    price INTEGER,
    discount FLOAT,
    avg_rating FLOAT,
    total_ratings INTEGER
);
"""
cursor.execute(create_table_query)

# -------------------------------
# 4. Insert Data using Pandas
# -------------------------------
df.to_sql("product", conn, if_exists="append", index=False)

# -------------------------------
# 5. Verify Insert (Optional)
# -------------------------------
check_df = pd.read_sql("SELECT * FROM product LIMIT 5", conn)
print("Sample data from SQLite:")
print(check_df)

# -------------------------------
# 6. Close Connection
# -------------------------------
conn.commit()
conn.close()

print("✅ Data successfully inserted into SQLite database: db.sqlite (table: product)")
```
